# Sample Tool-Usage Analysis

This notebook demonstrates a small, dependency-free analysis of tool frequency and exact prefix-cache hits.

```python
import csv
from collections import Counter
from pathlib import Path

data_path = Path("sample_tool_usage.csv")
with data_path.open(newline="", encoding="utf-8") as csv_file:
    rows = list(csv.DictReader(csv_file))

rows[:3]
```

```python
tool_frequency = Counter(row["tool_name"] for row in rows)
tool_frequency
```

```python
cache_summary = {}
for tool, calls in tool_frequency.items():
    tool_rows = [row for row in rows if row["tool_name"] == tool]
    hits = sum(row["cache_hit"].lower() == "true" for row in tool_rows)
    cache_summary[tool] = {
        "calls": calls,
        "cache_hit_rate": hits / calls,
        "average_prefill_ms": sum(float(row["prefill_ms"]) for row in tool_rows) / calls,
    }

cache_summary
```
